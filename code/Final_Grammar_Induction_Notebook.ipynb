## Data and split

I shuffle the 3,914 trees with seed 42, take the first 1,305 for Partition A, and shuffle that partition again before splitting it. I remove empty categories and strip functional suffixes such as NP-SBJ. The words, case and punctuation stay in the model input.

```python
def clean_tree(node):
    if isinstance(node, str):
        return node
    if node.label() == "-NONE-":
        return None

    children = [clean_tree(child) for child in node]
    children = [child for child in children if child is not None]
    if not children:
        return None

    label = node.label()
    if not label.startswith("-"):
        label = label.split("-", 1)[0].split("=", 1)[0]
    return Tree(label, children)


try:
    all_trees = list(treebank.parsed_sents())
except LookupError:
    nltk.download("treebank", quiet=True)
    all_trees = list(treebank.parsed_sents())

assert len(all_trees) == 3914
indices = np.random.RandomState(seed).permutation(len(all_trees))
partitions = {
    "A": indices[:1305].tolist(),
    "B": indices[1305:2610].tolist(),
    "C": indices[2610:].tolist(),
}
partition_a = np.random.RandomState(seed).permutation(partitions["A"])
train_end = int(0.8 * len(partition_a))
validation_end = train_end + int(0.1 * len(partition_a))
split_indices = {
    "train": partition_a[:train_end].tolist(),
    "validation": partition_a[train_end:validation_end].tolist(),
    "test": partition_a[validation_end:].tolist(),
}
trees_by_split = {
    split: [clean_tree(all_trees[index]) for index in tree_indices]
    for split, tree_indices in split_indices.items()
}
train_trees = trees_by_split["train"]
validation_trees = trees_by_split["validation"]
test_trees = trees_by_split["test"]

save_json("partition_indices.json", {"partitions": partitions, "partition_a_splits": split_indices})
split_text = json.dumps(split_indices, sort_keys=True, separators=(",", ":"))
run_info["split_sha256"] = hashlib.sha256(split_text.encode()).hexdigest()
display(pd.DataFrame([
    {"Split": split.title(), "Trees": len(trees)} for split, trees in trees_by_split.items()
]))
```

```python
def replace_rare_words(trees, cutoff):
    word_counts = Counter(word for tree in trees for word in tree.leaves())
    changed_trees = []
    for tree in trees:
        changed = tree.copy(deep=True)
        for position in changed.treepositions("leaves"):
            if word_counts[changed[position]] <= cutoff:
                changed[position] = "<UNK>"
        changed_trees.append(changed)
    return changed_trees


def to_cnf(tree, markov_order):
    binary_tree = tree.copy(deep=True)
    binary_tree.collapse_unary(collapsePOS=True, collapseRoot=True)
    binary_tree.chomsky_normal_form(factor="right", horzMarkov=markov_order)
    return binary_tree
```

## How I score the parses

A bracket records where a phrase starts and ends. Labeled F1 also checks its category, such as NP. The original Task 1 scores keep punctuation. For the comparison across all three tasks, I remove punctuation using the gold POS tags, count each remaining span once, and exclude single words and the full sentence. A sentence with fewer than two content words is skipped. A failed parse on an eligible sentence still counts against recall.

```python
punctuation_tags = {".", ",", ":", "``", "''", "-LRB-", "-RRB-"}


def brackets(tree, labeled=False):
    spans = Counter()

    def visit(node, start, is_root=False):
        if isinstance(node, str):
            return start + 1
        end = start
        for child in node:
            end = visit(child, end)
        if not is_root and any(isinstance(child, Tree) for child in node):
            span = (start, end, node.label()) if labeled else (start, end)
            spans[span] += 1
        return end

    if tree is not None:
        visit(tree, 0, is_root=True)
    return spans


def content_length(tree):
    return sum(tag not in punctuation_tags for _, tag in tree.pos())


def project_spans(spans, prefix):
    # prefix[i] counts the content words before original position i.
    projected = set()
    for start, end in spans:
        left, right = prefix[start], prefix[end]
        if 2 <= right - left < prefix[-1]:
            projected.add((left, right))
    return projected


def score_sentence(split, index, gold_tree, prediction, predicted_spans=None, one_sided=None):
    words = gold_tree.leaves()
    gold_tags = gold_tree.pos()
    prefix = [0]
    for _, tag in gold_tags:
        prefix.append(prefix[-1] + int(tag not in punctuation_tags))

    eligible = prefix[-1] >= 2
    if prediction is not None:
        assert prediction.leaves() == words

    gold = project_spans(brackets(gold_tree), prefix)
    predicted = project_spans(
        brackets(prediction) if predicted_spans is None else predicted_spans, prefix
    )
    matches = gold & predicted
    span_count = len(gold) + len(predicted)
    sentence_f1 = 2 * len(matches) / span_count if span_count else 1.0

    return {
        "position": index, "corpus_index": split_indices[split][index],
        "tokens": words,
        "scored_words": [word for word, tag in gold_tags if tag not in punctuation_tags],
        "eligible": eligible, "complete": prediction is not None, "one_sided": one_sided,
        "matching": len(matches), "predicted_count": len(predicted), "gold_count": len(gold),
        "sentence_f1": sentence_f1 if eligible else None,
        "gold_spans": sorted(gold), "predicted_spans": sorted(predicted),
        "matching_spans": sorted(matches),
        "extra_spans": sorted(predicted - gold), "missed_spans": sorted(gold - predicted),
        "gold_tree": gold_tree.pformat(margin=10**9),
        "predicted_tree": prediction.pformat(margin=10**9) if prediction is not None else None,
    }


def bracket_scores(matches, predicted, gold):
    precision = matches / predicted if predicted else 0.0
    recall = matches / gold if gold else 0.0
    f1 = 2 * matches / (predicted + gold) if predicted + gold else 0.0
    return precision, recall, f1


def save_evaluation(name, records):
    scored = [row for row in records if row["eligible"]]
    matches = sum(row["matching"] for row in scored)
    predicted = sum(row["predicted_count"] for row in scored)
    gold = sum(row["gold_count"] for row in scored)
    precision, recall, corpus_f1 = bracket_scores(matches, predicted, gold)
    diagnosed = [row["one_sided"] for row in scored if row["one_sided"] is not None]
    summary = {
        "sentences": len(records), "scored": len(scored),
        "skipped": len(records) - len(scored),
        "complete": sum(row["complete"] for row in scored),
        "matching_spans": matches, "predicted_spans": predicted, "gold_spans": gold,
        "precision": precision, "recall": recall, "corpus_f1": corpus_f1,
        "sentence_f1": sum(row["sentence_f1"] for row in scored) / len(scored),
        "one_sided": sum(diagnosed), "long_sentences_diagnosed": len(diagnosed),
    }
    save_json(f"{name}_metrics.json", summary)
    with (run_dir / f"{name}_predictions.jsonl").open("w", encoding="utf-8") as file:
        for row in records:
            file.write(json.dumps(row, ensure_ascii=False) + "\n")
    return summary


def score_row(name, result):
    return {
        "Model": name,
        "Scored": result["scored"],
        "Complete parses": result["complete"],
        "Precision (%)": 100 * result["precision"],
        "Recall (%)": 100 * result["recall"],
        "Sentence F1 (%)": 100 * result["sentence_f1"],
        "Corpus F1 (%)": 100 * result["corpus_f1"],
    }


def show_examples(records, task):
    examples = [row for row in records if row["eligible"] and row["gold_count"]
                and 5 <= len(row["tokens"]) <= 25]
    if len(examples) < 2:
        return
    better = max(examples, key=lambda row: row["sentence_f1"])
    worse = min((row for row in examples if row is not better),
                key=lambda row: row["sentence_f1"])

    for name, row in [("High F1", better), ("Low F1", worse)]:
        print(f"{name} · test sentence {row['position']} · F1 {row['sentence_f1']:.1%}")
        print(" ".join(row["tokens"]))
        print("Gold:     ", Tree.fromstring(row["gold_tree"]).pformat(margin=95))
        predicted = row["predicted_tree"]
        print("Predicted:", Tree.fromstring(predicted).pformat(margin=95) if predicted else "No parse")
        words = row["scored_words"]
        for label, key in [("Matched", "matching_spans"), ("Missed", "missed_spans"),
                           ("Extra", "extra_spans")]:
            phrases = [" ".join(words[i:j]) for i, j in row[key]]
            print(f"{label}: {', '.join(phrases) or '—'}")
        print()

    save_json(f"{task}_examples.json", [better, worse])
```

## Task 1: Counted PCFG and CKY

I count rules after collapsing unary chains and binarizing the trees. Counts below 5 get one extra count before I normalize by the parent. I try unknown word cutoffs 1 and 2 and select the cutoff for each Markov order using labeled validation F1.

```python
def fit_counted_pcfg(cnf_trees):
    rule_counts = Counter(rule for tree in cnf_trees for rule in tree.productions())
    smoothed = {rule: count + int(count < 5) for rule, count in rule_counts.items()}
    parent_totals = Counter()
    for rule, count in smoothed.items():
        parent_totals[rule.lhs()] += count

    lexical_rules = defaultdict(list)
    binary_rules = defaultdict(lambda: defaultdict(list))
    for rule, count in smoothed.items():
        parent = rule.lhs().symbol()
        log_prob = log(count / parent_totals[rule.lhs()])
        children = rule.rhs()
        if len(children) == 1 and isinstance(children[0], str):
            lexical_rules[children[0]].append((parent, log_prob))
        elif len(children) == 2:
            left, right = (child.symbol() for child in children)
            binary_rules[left][right].append((parent, log_prob))
        else:
            raise ValueError(f"Rule is outside CNF: {rule}")

    root_counts = Counter(tree.label() for tree in cnf_trees)
    root_scores = {root: log(count / len(cnf_trees)) for root, count in root_counts.items()}
    return lexical_rules, binary_rules, root_scores
```

### CKY decoding

Each chart cell covers a span. For each label, it stores the best log score and the split that gave that score. Once the chart covers the full sentence, I add the root score and follow the stored splits to rebuild the tree.

```python
def parse_counted(words, grammar):
    lexical_rules, binary_rules, root_scores = grammar
    n = len(words)
    if n == 0:
        return None
    chart = [[{} for _ in range(n + 1)] for _ in range(n)]
    label_order = [[{} for _ in range(n + 1)] for _ in range(n)]

    for start, word in enumerate(words):
        token = word if word in lexical_rules else "<UNK>"
        cell = chart[start][start + 1]
        for label, score in lexical_rules.get(token, ()):
            cell[label] = (score, None)
        label_order[start][start + 1] = {label: i for i, label in enumerate(cell)}

    for width in range(2, n + 1):
        for start in range(n - width + 1):
            end = start + width
            cell = chart[start][end]
            for split in range(start + 1, end):
                left_cell, right_cell = chart[start][split], chart[split][end]
                if not left_cell or not right_cell:
                    continue
                right_order = label_order[split][end]
                for left, (left_score, _) in left_cell.items():
                    rules = binary_rules.get(left)
                    if not rules:
                        continue
                    # Only try child pairs with a rule. Keep chart order for ties.
                    for right in sorted(right_cell.keys() & rules.keys(), key=right_order.get):
                        child_score = left_score + right_cell[right][0]
                        for parent, rule_score in rules[right]:
                            score = child_score + rule_score
                            if parent not in cell or score > cell[parent][0]:
                                cell[parent] = (score, (split, left, right))
            label_order[start][end] = {label: i for i, label in enumerate(cell)}

    root = None
    best_score = float("-inf")
    for label, root_score in root_scores.items():
        if label in chart[0][n]:
            score = chart[0][n][label][0] + root_score
            if score > best_score:
                root, best_score = label, score
    if root is None:
        return None

    def build_tree(start, end, label):
        if end - start == 1:
            return Tree(label, [words[start]])
        split, left, right = chart[start][end][label][1]
        return Tree(label, [build_tree(start, split, left), build_tree(split, end, right)])

    tree = build_tree(0, n, root)
    tree.un_chomsky_normal_form()
    return tree
```

```python
train_words = {word for tree in train_trees for word in tree.leaves()}


def evaluate_counted(gold_trees, grammar):
    predictions = [parse_counted(tree.leaves(), grammar) for tree in gold_trees]
    lexical_rules, _, _ = grammar
    results = {
        "sentences": len(gold_trees),
        "unseen": sum(any(word not in train_words for word in tree.leaves()) for tree in gold_trees),
        "using_unk": sum(any(word not in lexical_rules for word in tree.leaves()) for tree in gold_trees),
        "complete": sum(tree is not None for tree in predictions),
    }
    for name, labeled in [("Labeled", True), ("Unlabeled", False)]:
        matches = predicted_count = gold_count = 0
        for prediction, gold_tree in zip(predictions, gold_trees):
            gold = brackets(gold_tree, labeled)
            predicted = brackets(prediction, labeled)
            matches += sum((gold & predicted).values())
            predicted_count += sum(predicted.values())
            gold_count += sum(gold.values())
        results[name] = bracket_scores(matches, predicted_count, gold_count)
    return results, predictions
```

```python
task1_results = {}
task1_records = {}
validation_rows = []
test_rows = []
rare_word_trees = {cutoff: replace_rare_words(train_trees, cutoff) for cutoff in (1, 2)}

for order in (0, 1):
    grammars, validation_results = {}, {}
    for cutoff in (1, 2):
        training_cnf = [to_cnf(tree, order) for tree in rare_word_trees[cutoff]]
        grammars[cutoff] = fit_counted_pcfg(training_cnf)
        validation_results[cutoff], _ = evaluate_counted(validation_trees, grammars[cutoff])
        print(f"Finished validation: order {order}, cutoff {cutoff}")

    best_cutoff = max(validation_results, key=lambda cutoff: validation_results[cutoff]["Labeled"][2])
    test_result, predictions = evaluate_counted(test_trees, grammars[best_cutoff])
    for cutoff, result in validation_results.items():
        validation_rows.append({
            "Order": order, "Cutoff": cutoff, "Selected": cutoff == best_cutoff,
            "Labeled validation F1 (%)": 100 * result["Labeled"][2],
            "Unlabeled validation F1 (%)": 100 * result["Unlabeled"][2],
        })
    test_rows.append({
        "Order": order, "Cutoff": best_cutoff, "Complete parses": test_result["complete"],
        "Labeled test F1 (%)": 100 * test_result["Labeled"][2],
        "Unlabeled test F1 (%)": 100 * test_result["Unlabeled"][2],
    })
    records = [
        score_sentence("test", index, gold, prediction)
        for index, (gold, prediction) in enumerate(zip(test_trees, predictions))
    ]
    task1_records[order] = records
    task1_results[order] = save_evaluation(f"task1_h{order}_test", records)
    save_json(f"task1_h{order}_original_metrics.json", {
        "markov_order": order, "selected_cutoff": best_cutoff,
        "validation_original": validation_results[best_cutoff], "test_original": test_result,
        "aligned_unlabeled_test": task1_results[order],
    })

display(pd.DataFrame(validation_rows).round(2))
display(pd.DataFrame(test_rows).round(2))
```

## Task 2: Compound PCFG

This model sees only the words during training. A bidirectional LSTM turns each sentence into a 64-dimensional latent vector z, and the grammar predicts rule probabilities conditioned on z. I leave out the one single-token training sentence because the binary inside calculation needs at least two words. Batching sentences of equal length avoids padding.

```python
def encode_words(words, vocabulary):
    return [vocabulary.get(word, vocabulary["<UNK>"]) for word in words]


def length_batches(sentences, batch_size=4):
    by_length = defaultdict(list)
    for sentence in sentences:
        by_length[len(sentence)].append(sentence)
    batches = []
    for length in sorted(by_length):
        group = by_length[length]
        for start in range(0, len(group), batch_size):
            batches.append(torch.tensor(group[start:start + batch_size], dtype=torch.long))
    return batches


training_sentences = [tree.leaves() for tree in train_trees if len(tree.leaves()) >= 2]
word_counts = Counter(word for sentence in training_sentences for word in sentence)
ranked_words = sorted(word_counts, key=lambda word: (-word_counts[word], word))
compound_vocab = {"<UNK>": 0}
for word in ranked_words[:4000]:
    compound_vocab[word] = len(compound_vocab)

train_batches = length_batches([encode_words(words, compound_vocab) for words in training_sentences])
validation_batches = length_batches([
    encode_words(tree.leaves(), compound_vocab) for tree in validation_trees
])
print(f"{len(training_sentences):,} training sentences, {len(compound_vocab):,} word IDs.")
```

### Encoder and rule probabilities

The encoder predicts a mean and log variance for z. The grammar uses three MLPs: one for root labels, one for pairs of child labels, and one for word emissions. NT means nonterminal; PT means preterminal, the label directly above a word.

```python
def rule_mlp(input_size, output_size):
    return nn.Sequential(nn.Linear(input_size, 256), nn.ReLU(), nn.Linear(256, output_size))


class SentenceEncoder(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, 256)
        self.bilstm = nn.LSTM(256, 256, num_layers=1, bidirectional=True, batch_first=True)
        self.to_params = nn.Linear(512, 128)

    def forward(self, word_ids):
        word_vectors = self.embedding(word_ids)
        hidden_states, _ = self.bilstm(word_vectors)
        sentence_vector = hidden_states.max(dim=1).values
        return self.to_params(sentence_vector).chunk(2, dim=-1)


class CompoundGrammar(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.num_nt = 30
        self.num_pt = 60
        self.num_symbols = self.num_nt + self.num_pt
        self.root_embedding = nn.Parameter(torch.randn(256) * 0.02)
        self.nt_embedding = nn.Embedding(self.num_nt, 256)
        self.pt_embedding = nn.Embedding(self.num_pt, 256)

        self.root_mlp = rule_mlp(256 + 64, self.num_nt)
        self.rule_mlp = rule_mlp(256 + 64, self.num_symbols ** 2)
        self.word_mlp = rule_mlp(256 + 64, vocab_size)

    def forward(self, z):
        batch_size = z.size(0)
        root_input = torch.cat([self.root_embedding.expand(batch_size, -1), z], dim=-1)
        root_scores = F.log_softmax(self.root_mlp(root_input), dim=-1)

        nt = self.nt_embedding.weight.unsqueeze(0).expand(batch_size, -1, -1)
        nt_input = torch.cat([nt, z.unsqueeze(1).expand(-1, self.num_nt, -1)], dim=-1)
        rule_scores = F.log_softmax(self.rule_mlp(nt_input), dim=-1)
        rule_scores = rule_scores.view(batch_size, self.num_nt, self.num_symbols, self.num_symbols)

        pt = self.pt_embedding.weight.unsqueeze(0).expand(batch_size, -1, -1)
        pt_input = torch.cat([pt, z.unsqueeze(1).expand(-1, self.num_pt, -1)], dim=-1)
        word_scores = F.log_softmax(self.word_mlp(pt_input), dim=-1)
        return root_scores, rule_scores, word_scores


torch.manual_seed(seed)
encoder = SentenceEncoder(len(compound_vocab)).to(device)
compound_grammar = CompoundGrammar(len(compound_vocab)).to(device)
```

### Training objective

The inside algorithm sums the probabilities of all possible trees for the sentence. I minimize the negative log of that sum plus the KL penalty for z. When training a fresh model, I keep the checkpoint with the lowest validation loss.

```python
def sentence_distribution(word_ids, z):
    root_scores, rule_scores, word_scores = compound_grammar(z)
    # For each preterminal, select the scores of the words in the input.
    word_indices = word_ids.unsqueeze(1).expand(-1, word_scores.size(1), -1)
    emissions = word_scores.gather(2, word_indices).transpose(1, 2).contiguous()
    return SentCFG((emissions, rule_scores, root_scores))


def compound_loss(word_ids):
    mean, log_variance = encoder(word_ids)
    std = torch.exp(0.5 * log_variance)
    z = mean + torch.randn_like(std) * std
    negative_log_likelihood = -sentence_distribution(word_ids, z).partition
    kl = -0.5 * (1 + log_variance - mean.square() - log_variance.exp()).sum(dim=1)
    return negative_log_likelihood + kl


@torch.no_grad()
def compound_validation_loss():
    encoder.eval()
    compound_grammar.eval()
    loss_sum = sentence_count = 0
    gpu_devices = [torch.cuda.current_device()] if device.type == "cuda" else []
    with torch.random.fork_rng(devices=gpu_devices):
        # Use the same noise samples when comparing epochs.
        torch.manual_seed(1234)
        for batch in validation_batches:
            loss_sum += compound_loss(batch.to(device)).sum().item()
            sentence_count += len(batch)
    return loss_sum / sentence_count


def train_compound(epochs=15):
    parameters = list(encoder.parameters()) + list(compound_grammar.parameters())
    optimizer = torch.optim.Adam(parameters, lr=1e-3, betas=(0.75, 0.999))
    best_loss = float("inf")
    best_path = new_model_dir / "task2_best.pt"
    history = []

    for epoch in range(1, epochs + 1):
        encoder.train()
        compound_grammar.train()
        batch_order = list(range(len(train_batches)))
        random.Random(seed + epoch).shuffle(batch_order)
        loss_sum = sentence_count = 0

        for batch_index in batch_order:
            batch = train_batches[batch_index].to(device)
            optimizer.zero_grad(set_to_none=True)
            loss = compound_loss(batch).mean()
            if not torch.isfinite(loss):
                raise ValueError(f"Invalid Task 2 loss in epoch {epoch}")
            loss.backward()
            torch.nn.utils.clip_grad_norm_(parameters, 3.0)
            optimizer.step()

            loss_sum += loss.item() * len(batch)
            sentence_count += len(batch)

        train_loss = loss_sum / sentence_count
        validation_loss = compound_validation_loss()
        improved = validation_loss < best_loss
        if improved:
            best_loss = validation_loss
        checkpoint = {
            "epoch": epoch, "encoder": encoder.state_dict(),
            "grammar": compound_grammar.state_dict(), "optimizer": optimizer.state_dict(),
            "vocab": compound_vocab, "best_validation_loss": best_loss,
        }
        torch.save(checkpoint, new_model_dir / "task2_latest.pt")
        if improved:
            torch.save(checkpoint, best_path)
        history.append({"Epoch": epoch, "Train": train_loss, "Validation": validation_loss})
        print(f"Epoch {epoch:2d}: train {train_loss:.3f}, validation {validation_loss:.3f}")

    loss_table = pd.DataFrame(history)
    loss_table.to_csv(run_dir / "task2_training_loss.csv", index=False)
    ax = loss_table.plot(x="Epoch", marker="o", ylabel="Negative ELBO", title="Task 2 training loss")
    ax.figure.tight_layout()
    ax.figure.savefig(run_dir / "task2_training_loss.png", dpi=160)
    plt.show()
    return best_path
```

### Choose the decoder on validation

Viterbi returns the most probable tree. Minimum Bayes risk (MBR) uses the posterior probability of each bracket and chooses a tree whose brackets have the highest total probability. Both use the encoder mean for z at test time.

```python
@torch.no_grad()
def decode_distribution(sentence_ids):
    word_ids = torch.tensor([sentence_ids], dtype=torch.long, device=device)
    mean, _ = encoder(word_ids)
    return sentence_distribution(word_ids, mean)


def viterbi_spans(sentence_ids):
    with torch.enable_grad():
        span_mask = decode_distribution(sentence_ids).argmax[3][0].detach()

    n = len(sentence_ids)
    spans = Counter()
    for width_index, start, _ in (span_mask > 0.5).nonzero(as_tuple=False).tolist():
        end = start + width_index + 2
        if (start, end) != (0, n):
            spans[start, end] += 1
    return spans


def mbr_spans(sentence_ids):
    with torch.enable_grad():
        marginals = decode_distribution(sentence_ids).marginals[3][0]
    # Sum over labels; Task 2 evaluates unlabeled spans.
    span_scores = marginals.sum(dim=-1).detach().cpu()

    n = len(sentence_ids)
    best = [[0.0] * (n + 1) for _ in range(n + 1)]
    splits = [[None] * (n + 1) for _ in range(n + 1)]
    for width in range(2, n + 1):
        for start in range(n - width + 1):
            end = start + width
            split = max(range(start + 1, end), key=lambda k: best[start][k] + best[k][end])
            reward = 0.0 if width == n else span_scores[width - 2, start].item()
            best[start][end] = best[start][split] + best[split][end] + reward
            splits[start][end] = split

    spans = Counter()

    def collect(start, end):
        if end - start < 2:
            return
        if (start, end) != (0, n):
            spans[start, end] += 1
        split = splits[start][end]
        collect(start, split)
        collect(split, end)

    collect(0, n)
    return spans


def unlabeled_tree(words, spans):
    # X and W are display labels for internal nodes and words.
    spans = set(spans)

    def build(start, end):
        if end - start == 1:
            return Tree("W", [words[start]])
        for split in range(start + 1, end):
            left_ok = split == start + 1 or (start, split) in spans
            right_ok = end == split + 1 or (split, end) in spans
            if left_ok and right_ok:
                return Tree("X", [build(start, split), build(split, end)])
        raise ValueError(f"Cannot rebuild span {(start, end)}")

    return build(0, len(words))


def evaluate_compound(split, decoder_name, decode):
    records = []
    for index, gold in enumerate(trees_by_split[split]):
        words = gold.leaves()
        if content_length(gold) < 2:
            records.append(score_sentence(split, index, gold, None))
            continue

        spans = decode(encode_words(words, compound_vocab))
        prediction = unlabeled_tree(words, spans)
        one_sided = None
        if len(words) >= 8:
            right_ends = Counter()
            for (_, end), count in spans.items():
                right_ends[end] += count
            one_sided = max(right_ends.values()) / sum(right_ends.values()) >= 0.8
        records.append(score_sentence(split, index, gold, prediction, spans, one_sided))

    result = save_evaluation(f"task2_{split}_{decoder_name.lower()}", records)
    return result, records
```

```python
decoders = {"Viterbi": viterbi_spans, "MBR": mbr_spans}
task2_validation = {}
for name, decode in decoders.items():
    task2_validation[name], _ = evaluate_compound("validation", name, decode)

chosen_decoder = max(task2_validation, key=lambda name: task2_validation[name]["sentence_f1"])
decoder_table = pd.DataFrame([
    score_row(name, result) for name, result in task2_validation.items()
]).rename(columns={"Model": "Decoder"})
display(decoder_table.round(2))
print("Selected on validation sentence F1:", chosen_decoder)

ax = decoder_table.set_index("Decoder")[["Sentence F1 (%)", "Corpus F1 (%)"]].plot.bar(
    color=["#3768a2", "#278577"], rot=0, figsize=(7, 3.7), ylim=(0, 100),
    ylabel="F1 (%)", title="Task 2 · validation decoder comparison",
)
for bars in ax.containers:
    ax.bar_label(bars, fmt="%.2f", padding=3, fontsize=9)
ax.legend(loc="upper right")
ax.figure.tight_layout()
ax.figure.savefig(run_dir / "task2_validation_decoders.png", dpi=160)
plt.show()
```

## Task 3: Supervised neural PCFG

Here the training trees already give the category labels and the rule used at each node. I use the same embedding and MLP design as Task 2 without z. The labels and allowed binary rules come from the gold training trees, using Markov order 1 and the singleton word cutoff.

```python
supervised_trees = [to_cnf(tree, markov_order=1) for tree in rare_word_trees[1]]
observed_rules = set()
preterminals = set()
for tree in supervised_trees:
    for rule in tree.productions():
        parent = rule.lhs().symbol()
        if len(rule.rhs()) == 2:
            left, right = (child.symbol() for child in rule.rhs())
            observed_rules.add((parent, left, right))
        else:
            preterminals.add(parent)

nonterminal_labels = sorted({parent for parent, _, _ in observed_rules})
preterminal_labels = sorted(preterminals)
nonterminal_id = {label: i for i, label in enumerate(nonterminal_labels)}
preterminal_id = {label: i for i, label in enumerate(preterminal_labels)}
root_labels = sorted({tree.label() for tree in supervised_trees})
root_id = {label: i for i, label in enumerate(root_labels)}
child_pairs = sorted({(left, right) for _, left, right in observed_rules})
pair_id = {pair: i for i, pair in enumerate(child_pairs)}

words = {word for tree in supervised_trees for word in tree.leaves()}
word_list = ["<UNK>"] + sorted(words - {"<UNK>"})
supervised_vocab = {word: i for i, word in enumerate(word_list)}

allowed_pairs = torch.zeros(len(nonterminal_labels), len(child_pairs), dtype=torch.bool)
for parent, left, right in observed_rules:
    allowed_pairs[nonterminal_id[parent], pair_id[left, right]] = True

display(pd.DataFrame({
    "Training trees": [len(supervised_trees)],
    "Nonterminals": [len(nonterminal_labels)],
    "Preterminals": [len(preterminal_labels)],
    "Binary rules": [len(observed_rules)],
    "Root labels": [len(root_labels)],
    "Word IDs": [len(word_list)],
}))
```

### Score a gold tree

I add the log probabilities of its root, binary rules and word emissions. With the labels and splits already given, the training loss is the negative score of that one derivation.

```python
class SupervisedGrammar(nn.Module):
    def __init__(self):
        super().__init__()
        self.nt_embedding = nn.Embedding(len(nonterminal_labels), 256)
        self.pt_embedding = nn.Embedding(len(preterminal_labels), 256)
        self.root_embedding = nn.Parameter(torch.randn(256) * 0.02)
        self.root_mlp = rule_mlp(256, len(root_labels))
        self.rule_mlp = rule_mlp(256, len(child_pairs))
        self.word_mlp = rule_mlp(256, len(word_list))
        self.register_buffer("allowed", allowed_pairs.clone())

    def forward(self):
        root_scores = F.log_softmax(self.root_mlp(self.root_embedding), dim=-1)
        binary = self.rule_mlp(self.nt_embedding.weight)
        binary = binary.masked_fill(~self.allowed, float("-inf"))
        rule_scores = F.log_softmax(binary, dim=-1)
        word_scores = F.log_softmax(self.word_mlp(self.pt_embedding.weight), dim=-1)
        return root_scores, rule_scores, word_scores


def gold_tree_score(tree, root_scores, rule_scores, word_scores):
    def node_score(node):
        label = node.label()
        if isinstance(node[0], str):
            return word_scores[preterminal_id[label], supervised_vocab[node[0]]]

        left, right = node
        pair = pair_id[left.label(), right.label()]
        return rule_scores[nonterminal_id[label], pair] + node_score(left) + node_score(right)

    return root_scores[root_id[tree.label()]] + node_score(tree)


torch.manual_seed(seed)
supervised_grammar = SupervisedGrammar().to(device)
label_mappings = {
    "nt_labels": nonterminal_labels, "pt_labels": preterminal_labels,
    "rhs_pairs": child_pairs, "root_labels": root_labels, "vocab": word_list,
}
```

```python
def train_supervised(epochs=15, batch_size=8):
    optimizer = torch.optim.Adam(supervised_grammar.parameters(), lr=1e-3, betas=(0.75, 0.999))
    for epoch in range(1, epochs + 1):
        supervised_grammar.train()
        if epoch == 1:
            order = list(range(len(supervised_trees)))
            random.Random(seed).shuffle(order)
        else:
            generator = torch.Generator().manual_seed(seed + epoch)
            order = torch.randperm(len(supervised_trees), generator=generator).tolist()

        loss_sum = 0.0
        for start in range(0, len(order), batch_size):
            batch = order[start:start + batch_size]
            scores = supervised_grammar()
            loss = torch.stack([-gold_tree_score(supervised_trees[i], *scores) for i in batch]).mean()
            if not torch.isfinite(loss):
                raise ValueError(f"Invalid Task 3 loss in epoch {epoch}")
            optimizer.zero_grad(set_to_none=True)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(supervised_grammar.parameters(), 3.0)
            optimizer.step()
            loss_sum += loss.item() * len(batch)
        print(f"Epoch {epoch:2d}: loss per tree {loss_sum / len(order):.3f}")

    path = new_model_dir / "task3_epoch15.pt"
    torch.save({
        "epoch": epochs, "model_state": supervised_grammar.state_dict(),
        "optimizer_state": optimizer.state_dict(), **label_mappings,
    }, path)
    return path


task3_path = train_supervised() if train_supervised_model else checkpoint_path("task3")
checkpoint = torch.load(task3_path, map_location="cpu", weights_only=True)
if checkpoint["epoch"] != 15:
    raise ValueError("Task 3 uses the fixed epoch-15 checkpoint.")
for key, expected in label_mappings.items():
    if key in checkpoint and json.dumps(checkpoint[key]) != json.dumps(expected):
        raise ValueError(f"Task 3 checkpoint has different {key}.")

state = checkpoint["model_state"]
if not torch.equal(state["allowed"].cpu(), allowed_pairs):
    raise ValueError("Task 3 checkpoint has a different set of allowed binary rules.")
supervised_grammar.load_state_dict(state)
supervised_grammar.eval()
record_checkpoint("task3", task3_path, checkpoint)
save_json("task3_mappings.json", label_mappings)
print(f"Task 3 checkpoint: {task3_path.name}, epoch {checkpoint['epoch']}")
del checkpoint, state
```

### CKY with the learned scores

For each rule and split, the score is left child + right child + rule. NumPy computes those candidates together. For each rule I keep its best split, then for each parent I keep its best rule. The backpointers recover the tree.

```python
with torch.no_grad():
    root_log, binary_log, word_log = (scores.cpu() for scores in supervised_grammar())

labels = nonterminal_labels + preterminal_labels
label_id = {label: i for i, label in enumerate(labels)}
root_ids = np.array([label_id[label] for label in root_labels])
root_log_scores = np.array([float(score) for score in root_log])
word_log_scores = word_log.numpy()

# Keep a fixed rule order for ties.
rules_by_children = defaultdict(list)
for parent, left, right in sorted(observed_rules):
    rules_by_children[left, right].append(parent)
cky_rules = [
    (parent, left, right)
    for (left, right), parents in rules_by_children.items()
    for parent in parents
]
parents, left_children, right_children = np.array([
    [label_id[parent], label_id[left], label_id[right]] for parent, left, right in cky_rules
]).T
binary_log_scores = np.array([
    binary_log[nonterminal_id[parent], pair_id[left, right]].item()
    for parent, left, right in cky_rules
], dtype=np.float32)
rule_numbers = np.arange(len(cky_rules))
```

```python
def parse_supervised(words, rule_scores=None):
    if rule_scores is None:
        rule_scores = binary_log_scores
    n = len(words)
    if n == 0:
        return None
    label_count = len(labels)
    chart = np.full((n, n + 1, label_count), -np.inf, dtype=np.float32)
    back_rule = np.full(chart.shape, -1, dtype=np.int32)
    back_split = np.full(chart.shape, -1, dtype=np.int32)

    word_ids = encode_words(words, supervised_vocab)
    for start, word_id in enumerate(word_ids):
        chart[start, start + 1, len(nonterminal_labels):] = word_log_scores[:, word_id]

    for width in range(2, n + 1):
        for start in range(n - width + 1):
            end = start + width
            splits = np.arange(start + 1, end)

            # One row per split, one column per rule.
            candidates = (
                chart[start, splits[:, None], left_children]
                + chart[splits[:, None], end, right_children]
                + rule_scores
            )
            best_split = candidates.argmax(axis=0)
            best_rule_score = candidates[best_split, rule_numbers]

            # Update the best score for each parent, even when its ID repeats.
            best_parent_score = chart[start, end]
            np.maximum.at(best_parent_score, parents, best_rule_score)
            winning = np.isfinite(best_rule_score) & (best_rule_score == best_parent_score[parents])
            best_rule = np.full(label_count, len(cky_rules))
            np.minimum.at(best_rule, parents[winning], rule_numbers[winning])
            valid = best_rule < len(cky_rules)
            back_rule[start, end, valid] = best_rule[valid]
            back_split[start, end, valid] = splits[best_split[best_rule[valid]]]

    full_scores = chart[0, n, root_ids] + root_log_scores
    best_root = full_scores.argmax()
    if not np.isfinite(full_scores[best_root]):
        return None

    def build_tree(start, end, label):
        if end - start == 1:
            return Tree(labels[label], [words[start]])
        rule = back_rule[start, end, label]
        split = back_split[start, end, label]
        return Tree(labels[label], [
            build_tree(start, split, left_children[rule]),
            build_tree(split, end, right_children[rule]),
        ])

    tree = build_tree(0, n, root_ids[best_root])
    tree.un_chomsky_normal_form()
    return tree
```

```python
def evaluate_supervised(split, rule_scores=None, name="task3"):
    records = []
    for index, gold in enumerate(trees_by_split[split]):
        prediction = None
        if content_length(gold) >= 2:
            prediction = parse_supervised(gold.leaves(), rule_scores)
        records.append(score_sentence(split, index, gold, prediction))
    return save_evaluation(f"{name}_{split}", records), records


task3_validation, _ = evaluate_supervised("validation")
task3_test, task3_records = evaluate_supervised("test")
display(pd.DataFrame([
    score_row("Validation", task3_validation), score_row("Test", task3_test)
]).rename(columns={"Model": "Split"}).round(2))
show_examples(task3_records, "task3")
```

## Extra experiment: Counted + Neural PCFG

I mixed the counted and neural binary rule probabilities to see whether training rule frequencies help the parser choose better trees. I chose the mix with the highest validation corpus F1, then scored that mix on the test split.

```python
# Match the counted and neural scores by (parent, left child, right child).
_, counted_binary, _ = fit_counted_pcfg(supervised_trees)
counted_by_rule = {}

for left, right_options in counted_binary.items():
    for right, rules in right_options.items():
        for parent, log_prob in rules:
            counted_by_rule[parent, left, right] = np.exp(log_prob)

counted_probs = np.array([counted_by_rule[rule] for rule in cky_rules])
neural_probs = np.exp(binary_log_scores.astype(float))


def mixed_rule_scores(weight):
    probs = (1 - weight) * neural_probs + weight * counted_probs
    return np.log(probs).astype(np.float32)


hybrid_validation = {0.0: task3_validation}
for weight in (0.25, 0.50):
    hybrid_validation[weight], _ = evaluate_supervised(
        "validation", mixed_rule_scores(weight), name=f"hybrid_{int(100 * weight)}"
    )

best_weight = max(hybrid_validation, key=lambda w: hybrid_validation[w]["corpus_f1"])

display(pd.DataFrame([
    {
        "Counted weight": f"{weight:.0%}",
        "Validation corpus F1 (%)": 100 * result["corpus_f1"],
        "Selected": weight == best_weight,
    }
    for weight, result in hybrid_validation.items()
]).round(2))

hybrid_test = task3_test
if best_weight > 0:
    hybrid_test, _ = evaluate_supervised(
        "test", mixed_rule_scores(best_weight), name="hybrid_selected"
    )

display(pd.DataFrame([
    score_row("Task 3 baseline", task3_test),
    score_row(f"Selected ({best_weight:.0%} counted)", hybrid_test),
]).round(2))

validation_change = 100 * (hybrid_validation[best_weight]["corpus_f1"] - task3_validation["corpus_f1"])
test_change = 100 * (hybrid_test["corpus_f1"] - task3_test["corpus_f1"])
print(f"Selected {best_weight:.0%} counted weight on validation.")
print(f"Corpus F1 change: {validation_change:+.2f} points on validation, {test_change:+.2f} points on test.")
save_json("hybrid_summary.json", {
    "selected_counted_weight": best_weight,
    "validation": hybrid_validation,
    "task3_baseline_test": task3_test,
    "selected_test": hybrid_test,
})
```

In the saved run, the mix improved validation F1 but lowered test F1. Both models parsed all 130 scored test sentences. Mixing changed the rule scores, so CKY could choose different brackets even when it could parse the same sentences. I kept the original neural PCFG in the main Task 3 comparison.
